```python
# 11
df = pd.read_csv('/content/IMDB-Movie-Data.csv')
df.groupby('Director')['Rating'].mean()
```

```python
# 12
df.groupby("Year")["Title"].agg(np.size)
```

```python
# 13
df[(df["Rating"] > 8) & (df["Revenue (Millions)"] > 100)]
```

```python
# 14
result = df.groupby("Director").agg({
    "Rating": "mean",
    "Revenue (Millions)": "mean"
})

result
```

```python
# 15
result = df.groupby("Year")[["Rating"]].agg(["count", "mean"])
result
```

```python
# 16 a)
df['Profitability'] = df['Revenue (Millions)'] / df['Runtime (Minutes)']
df[['Title', 'Revenue (Millions)', 'Runtime (Minutes)', 'Profitability']]
```

```python
# 16 b)
top_10 = df.sort_values('Profitability', ascending=False).head(10)
top_10[['Title', 'Profitability']]
```

```python
# 17 a)
df['Rating_Category'] = np.where(df['Rating'] < 6, 'Poor', '')
df[['Title', 'Rating', 'Rating_Category']]
```

```python
# 17 b)
df['Rating_Category'] = ''
df.loc[(df['Rating'] >= 6) & (df['Rating'] < 7.5), 'Rating_Category'] = "Average"
df[['Title', 'Rating', 'Rating_Category']]
```

```python
# 17 c)
df['Rating_Category'] = ''
df.loc[df['Rating'] >= 7.5, 'Rating_Category'] = 'Excellent'
df[['Title', 'Rating', 'Rating_Category']]
```

```python
#18 a)
def movie_score(row):
    return row['Rating'] * 10 + row['Revenue (Millions)']
df['High_Score'] = df.apply(movie_score, axis=1)
df[['Title', 'Rating', 'Revenue (Millions)', 'High_Score']]
```

```python
#18 b)
def score(row):
    return row['Rating'] * 10 + row['Revenue (Millions)']
df['Score'] = df.apply(score, axis=1)
df[['Title', 'Rating', 'Revenue (Millions)', 'Score']]
```
