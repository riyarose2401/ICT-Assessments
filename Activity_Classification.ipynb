```python
# Label Encoding

from sklearn.preprocessing import LabelEncoder

binary_cols = ['school', 'sex', 'address', 'famsize', 'Pstatus','schoolsup', 'famsup', 'paid', 'activities', 'nursery',
       'higher', 'internet', 'romantic','passed']
le = LabelEncoder()
for col in binary_cols:
    students_data[col] = le.fit_transform(students_data[col])
```

```python
students_data.info()
```

```python
students_data.head()
```

```python
# OneHot Encoding

students_data = pd.get_dummies(students_data,dtype='int',drop_first=True)
```

```python
students_data
```

```python
# Train test split
x = students_data.drop(['passed'],axis=1)
y = students_data['passed']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test =train_test_split(
    x,y,test_size=0.2,random_state=42
)
```

```python
# KNN
# Finding the optimum k value
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix,precision_score,accuracy_score,f1_score,recall_score
acc_score =[]
for i in range(3,10):
  knn = KNeighborsClassifier(n_neighbors = i)
  knn.fit(x_train,y_train)
  y_pred_knn = knn.predict(x_test)
  acc = accuracy_score(y_test,y_pred_knn)
  acc_score.append(acc)
```

```python
acc_score
```

```python
plt.plot(np.arange(3,10),acc_score,'o-')
plt.xlabel('k value')
plt.ylabel('Accuracy Score')
plt.grid()
```

```python
# optimum k values value is 5 because accuracy is high when k =5
# train the knn model k=5

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train,y_train)
y_pred_knn = knn.predict(x_test)
print(confusion_matrix(y_test,y_pred_knn))
print('Accuracy is : ',accuracy_score(y_test,y_pred_knn))
print('Precision is : ',precision_score(y_test,y_pred_knn))
print('Recall is : ',recall_score(y_test,y_pred_knn))
print('F1 score is : ',f1_score(y_test,y_pred_knn))
```

# Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,precision_score,accuracy_score,f1_score,recall_score
lr_model =LogisticRegression(max_iter=1000)
lr_model.fit(x_train,y_train)
y_pred_lr_model = lr_model.predict(x_test)
print(confusion_matrix(y_test,y_pred_lr_model))
print('Accuracy is: ',accuracy_score(y_test,y_pred_lr_model))
print('Precision is: ',precision_score(y_test,y_pred_lr_model))
print('Recall is: ',recall_score(y_test,y_pred_lr_model))
print('F1 score is: ',f1_score(y_test,y_pred_lr_model))
```
