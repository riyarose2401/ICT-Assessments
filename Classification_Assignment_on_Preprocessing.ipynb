# IQR method

```python
#Handling outlers in Age using IQR method

q1 = titanic_data['Age'].quantile(0.25)
q3 = titanic_data['Age'].quantile(0.75)

print(f'Q1 = {q1}\nQ3 = {q3}')

iqr = q3 - q1

up_lim_Age = q3 + (1.5 * iqr)
low_lim_Age = q1 - (1.5 * iqr)

print(f'IQR = {iqr}\nUp_lim = {up_lim_Age}\nLow_lim = {low_lim_Age}')
```

```python
outliers = titanic_data[(titanic_data['Age'] > up_lim_Age) | (titanic_data['Age'] < low_lim_Age)]
outliers
```

```python
titanic_data.loc[titanic_data['Age'] > up_lim_Age,'Age'] = up_lim_Age

titanic_data.loc[titanic_data['Age'] < low_lim_Age,'Age'] = low_lim_Age
```

```python
titanic_data['Age'].unique()
```

```python
#Handling outlers in Fare using IQR method

q1 = titanic_data['Fare'].quantile(0.25)
q2 = titanic_data['Fare'].quantile(0.5)
q3 = titanic_data['Fare'].quantile(0.75)
```

```python
print(f'Q1 = {q1}\nQ2 = {q2}\nQ3 = {q3}')
```

```python
iqr = q3 - q1
up_lim = q3+(1.5*iqr)
low_lim = q1 - (1.5*iqr)
print(f'IQR = {iqr}\nUp_lim = {up_lim}\nLow_lim = {low_lim}')
```

```python
titanic_data[titanic_data['Fare']<low_lim].index
```

```python
outliers =titanic_data[(titanic_data['Fare']>up_lim)|(
    titanic_data['Fare']<low_lim)
]
```

```python
#outliers =titanic[(titanic['Fare']>up_lim)|(
    #titanic['Fare']<low_lim)
#]

titanic_data.loc[titanic_data['Fare'] > up_lim, 'Fare'] = up_lim
titanic_data.loc[titanic_data['Fare'] < low_lim, 'Fare'] = low_lim
```

# Encoding

```python
# one hot encoding
titanic_data = pd.get_dummies(titanic_data,dtype='int',drop_first=True)
titanic_data
```

```python
# Feature Target Split

x=titanic_data.drop('Survived', axis= 1)
y=titanic_data['Survived']
```

```python
x.head(3)
```

```python
y.head(3)
```

```python
#MinMax Scaling
x1=x.copy()
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
for i in ['Age','Fare']:
  x1[i] = scaler.fit_transform(x1[[i]])
  x1.describe()
```

```python
# scaled
from sklearn.model_selection import train_test_split
x_train1, x_test1, y_train, y_test = train_test_split(x1, y, test_size=0.2, random_state=42, stratify=y)
```

```python
#non scaled x_train and x_test
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
```

# Logistic Regression

```python
from sklearn.metrics import confusion_matrix,precision_score,accuracy_score,f1_score,recall_score
```

```python
from sklearn.linear_model import LogisticRegression
lr_model =LogisticRegression(max_iter=1000)
lr_model.fit(x_train1,y_train)
y_pred_lr_model = lr_model.predict(x_test1)
print(confusion_matrix(y_test,y_pred_lr_model))
print('Accuracy is: ',accuracy_score(y_test,y_pred_lr_model))
print('Precision is: ',precision_score(y_test,y_pred_lr_model))
print('Recall is: ',recall_score(y_test,y_pred_lr_model))
print('F1 score is: ',f1_score(y_test,y_pred_lr_model))
```

# KNN

```python
# Finding the optimum k value
from sklearn.neighbors import KNeighborsClassifier

acc_score =[]
for i in range(3,10):
  knn = KNeighborsClassifier(n_neighbors = i)
  knn.fit(x_train1,y_train)
  y_pred_knn = knn.predict(x_test1)
  acc = accuracy_score(y_test,y_pred_knn)
  acc_score.append(acc)
```

```python
acc_score
```

```python
plt.plot(np.arange(3,10),acc_score,'o-')
plt.xlabel('k value')
plt.ylabel('Accuracy Score')
plt.grid()
```

```python
# optimum k value is 4 because accuracy is high when k = 4
knn= KNeighborsClassifier(n_neighbors=4)
knn.fit(x_train1,y_train)
y_pred_knn = knn.predict(x_test1)
print(confusion_matrix(y_test,y_pred_knn))
print('Accuracy is : ',accuracy_score(y_test,y_pred_knn))
print('Precision is : ',precision_score(y_test,y_pred_knn))
print('Recall is : ',recall_score(y_test,y_pred_knn))
print('F1 score is : ',f1_score(y_test,y_pred_knn))
```
