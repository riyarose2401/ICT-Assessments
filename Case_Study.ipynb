```python
#Question No:1
name = input("Enter account holder name: ")
account_no = input("Enter account number: ")
balance = 1000.0
def deposit(balance):
    amount = float(input("Enter deposit amount: "))

    if amount > 0:
        balance = balance + amount
        print("Deposit successful")
    else:
        print("Invalid amount")
   return balance
def withdrawal(balance):
    amount = float(input("Enter withdrawal amount: "))
if amount <= 0:
        print("Invalid amount")
    elif amount > balance:
        print("Insufficient balance")
    else:
        balance = balance - amount
        print("Withdrawal successful")
    return balance
def balance_enquiry(balance):
    print("Account Holder:", name)
    print("Account Number:", account_no)
    print("Balance:", balance)
while True:

    print("\n1. Deposit")
    print("2. Withdrawal")
    print("3. Balance Enquiry")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        balance = deposit(balance)

    elif choice == 2:
        balance = withdrawal(balance)

    elif choice == 3:
        balance_enquiry(balance)

    elif choice == 4:
        print("Thank you for using our bank!")
        break

    else:
        print("Invalid choice")



```

```python
# Question No:2

temperature = float(input('Enter temperature'))
humidity = float(input('Enter humidity'))
weather = input('Enter weather')
region = input("Enter the region: ")

temp1 = float(input('Enter temperature for day 1'))
temp2 = float(input('Enter temperature for day 2'))
temp3 = float(input('Enter temperature for day 3'))

humi_1 = float(input('Enter humidity for day 1'))
humi_2 = float(input('Enter humidity for day 2'))
humi_3 = float(input('Enter humidity for day 3'))

weather = input('Enter weather (sunny/rainy/cloudy)')
def average_temperature(t1, t2, t3):
    average = (t1 + t2 + t3) / 3
    return average

def average_humi(h1, h2, h3):
    average = (h1 + h2 + h3) / 3
    return average

def check_weather(temp):
    if temp >= 35:
        return 'Very Hot'
    elif temp >= 30:
        return 'Hot'
    elif temp >= 20:
        return 'Normal'
    else:
        return 'Cold'

avg_temp = average_temperature(temp1, temp2, temp3)
avg_humi = average_humi(humi_1, humi_2, humi_3)

condition = check_weather(avg_temp)

print('Weather Data Report')
print('Region:', region)
print('Weather:', weather)
print('Average Temperature:', avg_temp)
print('Average Humidity:', avg_humi)
print('Weather Condition:', condition)

```

```python
# Question No:3
def check_password(password):
  if len(password) < 8:
        return False
digit = False
    uppercase = False
    special = False
    for char in password:

        if char.isdigit():
            digit = True

        elif char.isupper():
            uppercase = True

        elif char in "!@#$%":
            special = True

    if digit and uppercase and special:
        return True
    else:
        return False


password = input('Enter your password')

if check_password(password):
    print('Strong password')
else:
    print('Weak password')
```

```python


#4
def most_frequent_words(text):
    text = text.lower()

    # Remove punctuation
    for symbol in ".,!?;:":
        text = text.replace(symbol, "")

    words = text.split()

    frequency = {}
    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1
```
